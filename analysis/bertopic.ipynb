```python
# Data processing
import pandas as pd
import numpy as np
# Visualization
import seaborn as sns
# Text preprocessiong
import nltk
nltk.download('stopwords')
# Dimension reduction
from umap import UMAP
# Clustering
from hdbscan import HDBSCAN
# Count vectorization
from sklearn.feature_extraction.text import CountVectorizer
# Topic model
from bertopic import BERTopic
from nltk.corpus import stopwords

class BertTopicModelling():
    def __init__(self) -> None:
        pass

    def read_file(self, path):
        df = pd.read_csv(path, header=None, names=['title'], sep='\t')

        return df


    def build_umap(self):
        return UMAP(n_neighbors=15, n_components=5, min_dist=0.0, metric='cosine', random_state=100)

    def build_cv(self, stop_words):
        return CountVectorizer(stop_words=stop_words)

    def build_model(self, ):
        umap_model = self.build_umap()
        stopwordssss = stopwords.words('slovene')
        vectorizer_model = self.build_cv(stop_words=stopwordssss)

        df = self.read_file('titles.csv')
        topic_model = BERTopic(umap_model=umap_model,
                       vectorizer_model=vectorizer_model,
                       # diversity=0.8,
                       min_topic_size=200,
                       top_n_words=4,
                       language="multilingual",
                       calculate_probabilities=True)

        print('done')
        topics = topic_model.fit_transform(df['title'])
        print('topic topic', topics)
        print(topic_model.get_topic_info())


    def main(self):
        self.build_model()

bt = BertTopicModelling()
bt.main()
```
